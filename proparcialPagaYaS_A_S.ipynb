```python
import csv

#contadoryListas

acumulable=0
pedidos=[]

trabajadores=["01250371062", "01250371059", "01250371084"]

ID=input("Por favor, Ingrese su ID de empleado: ")

if ID in trabajadores:

  print("Su ID ha sido validada")

  recaudo=int(input("Ingrese el tope maximo de recaudo: "))

  if recaudo>0:

    print("Su recaudo ha sido validado, prosiga")

    #atencionAlCliente

    while True:

      print("=====================================")
      print("1). Agregar")
      print("2). Leer")
      print("3). Actualizar")
      print("4). Borrar")
      print("0). Finalizar")
      print("=====================================")

      opcion=int(input("Ingrese su opcion: "))

      #agregar

      if opcion==1:

        monto=input("Ingrese el monto o FIN para terminar: ")

        if monto.upper()=="FIN" or monto=="0":

          break

        try:

          monto=int(monto)

          if monto<=0:

            print("El monto debe ser mayor que cero")

          else:

            pedidos.append(monto)
            acumulable=acumulable+monto

            print("Transaccion agregada correctamente")
            print("Recaudo acumulado:", acumulable)

            #suspensionPorSeguridad

            if acumulable>=recaudo:

              print("=====================================")
              print("CAJA SUSPENDIDA: se alcanzo el tope de recaudo.")
              print("Dirijase a tesoreria.")
              print("=====================================")

              break

        except ValueError:

          print("Monto no valido")

      #leer

      elif opcion==2:

        if len(pedidos)>0:

          for i in range(len(pedidos)):

            print(i+1, ".", pedidos[i])

        else:

          print("No hay transacciones")

      #actualizar

      elif opcion==3:

        if len(pedidos)>0:

          for i in range(len(pedidos)):

            print(i+1, ".", pedidos[i])

          posicion=int(input("Ingrese la transaccion a actualizar: "))

          if posicion>0 and posicion<=len(pedidos):

            nuevoMonto=int(input("Ingrese el nuevo monto: "))

            if nuevoMonto>0:

              acumulable=acumulable-pedidos[posicion-1]
              pedidos[posicion-1]=nuevoMonto
              acumulable=acumulable+nuevoMonto

              print("Transaccion actualizada")

            else:

              print("El monto debe ser mayor que cero")

          else:

            print("Transaccion no encontrada")

        else:

          print("No hay transacciones")

      #borrar

      elif opcion==4:

        if len(pedidos)>0:

          for i in range(len(pedidos)):

            print(i+1, ".", pedidos[i])

          posicion=int(input("Ingrese la transaccion a borrar: "))

          if posicion>0 and posicion<=len(pedidos):

            acumulable=acumulable-pedidos[posicion-1]
            pedidos.pop(posicion-1)

            print("Transaccion eliminada")

          else:

            print("Transaccion no encontrada")

        else:

          print("No hay transacciones")

      #finalizar

      elif opcion==0:

        break

      else:

        print("Opcion no valida")

    #crearCsv

    with open("recaudos.csv","w",newline="") as archivo:

      escritor=csv.writer(archivo)

      for monto in pedidos:

        escritor.writerow([ID,monto])

    #reporteFinal

    if len(pedidos)>0:

      promedio=acumulable/len(pedidos)

    else:

      promedio=0

    print("")
    print("=====================================")
    print("           REPORTE FINAL")
    print("=====================================")
    print("Cajero:", ID)
    print("Tope asignado:", recaudo)
    print("Transacciones procesadas:", len(pedidos))
    print("Recaudo total:", acumulable)
    print("Monto promedio:", round(promedio,2))
    print("=====================================")

  else:

    #recaudoMenorQueCero

    print("Su recaudo no es valido")

else:

  #idInvalida

  print("Su ID no es valida")
```
